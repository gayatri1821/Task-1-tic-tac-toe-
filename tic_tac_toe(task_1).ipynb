```python
board = [" " for _ in range(9)]
```

```python
def print_board(board):
    for i in range(0, 9, 3):
        print(board[i], "|", board[i+1], "|", board[i+2])
        if i < 6:
            print("---------")
```

```python
def check_winner(board, player):
    win_positions = [
        [0,1,2],[3,4,5],[6,7,8],
        [0,3,6],[1,4,7],[2,5,8],
        [0,4,8],[2,4,6]
    ]
    return any(all(board[i] == player for i in combo) for combo in win_positions)
```

```python
def is_draw(board):
    return " " not in board
```

```python
def minimax(board, depth, is_maximizing):
    if check_winner(board, "O"):
        return 1
    if check_winner(board, "X"):
        return -1
    if is_draw(board):
        return 0

    if is_maximizing:
        best = -100
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(board, depth+1, False)
                board[i] = " "
                best = max(score, best)
        return best
    else:
        best = 100
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                score = minimax(board, depth+1, True)
                board[i] = " "
                best = min(score, best)
        return best
```

```python
def best_move(board):
    best_score = -100
    move = 0

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(board, 0, False)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move
```

```python
while True:
    print_board(board)

    # Human move
    pos = int(input("Enter position (0-8): "))
    if board[pos] != " ":
        print("Invalid move")
        continue
    board[pos] = "X"

    if check_winner(board, "X"):
        print_board(board)
        print("You win!")
        break

    if is_draw(board):
        print("Draw!")
        break

    # AI move
    ai = best_move(board)
    board[ai] = "O"

    if check_winner(board, "O"):
        print_board(board)
        print("AI wins!")
        break

    if is_draw(board):
        print("Draw!")
        break
```
